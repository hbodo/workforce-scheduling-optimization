# Problème de planification

Une entreprise doit construire un planning hebdomadaire pour affecter ses employés à différents postes et créneaux, en respectant des contraintes de disponibilité, de compétences et de repos, tout en cherchant à produire un planning satisfaisant les préférences.

## 1. Données du problèmes

On considère une semaine de 7 jours, avec 3 créneaux par jour :

* matin
* après-midi
* nuit

Soit 21 shifts au total.

Chaque shift comporte plusieurs postes à pourvoir.

**Code des shifts :**
* M = matin : 06:00–14:00
* AP = après-midi : 14:00–22:00
* N = nuit : 22:00–06:00

### Contraintes qui doivent être respectées

Disponibilité : Un employé ne peut être affecté que sur un shift où il est disponible.

Une personne = un poste : Une personne ne peut occuper qu'un seul poste sur un shift.

Couverture : Tous les postes nécessaires doivent être couverts.

Compétences : Chaque shift doit avoir les compétences minimales requises.

Nombre maximal de shifts : Un employé ne peut pas dépasser son maximum hebdomadaire.

Repos : On interdit certaines successions.

| Succession | Autorisée ? |
| ---------- | ----------- |
| M → M      | oui         |
| M → AP      | oui         |
| M → N      | non         |
| AP → M      | non         |
| AP → A      | oui         |
| AP → N      | oui         |
| N → M      | non         |
| N → AP      | non         |
| N → N      | oui         |


### Contraintes souples :

On aimerait également :

* respecter les préférences des employés ;
* éviter les semaines trop chargées (6 shifts ou plus).

Mais ces contraintes peuvent être violées contre une pénalité.

Il faut noter également que les règles de départ sont simplifiées et non conformes au droit (journées de 16h possibles et semaines de 80h).
